```python
# Créer un dossier pour PlantDoc
PLANTDOC_DIR = '/content/plantdoc'
os.makedirs(PLANTDOC_DIR, exist_ok=True)

# Télécharger depuis Kaggle
# Plusieurs versions existent, on prend la plus standard
!kaggle datasets download -d nirmalsankalana/plantdoc-dataset -p {PLANTDOC_DIR} --unzip

# Voir ce qu'on a récupéré
!ls {PLANTDOC_DIR}
```

```python
# On regarde la structure
for root, dirs, files in os.walk(PLANTDOC_DIR):
    depth = root.replace(PLANTDOC_DIR, '').count(os.sep)
    if depth <= 2:
        indent = '  ' * depth
        print(f"{indent}{os.path.basename(root)}/")
```

```python
# Identifier le dossier "train" et lister ses classes
# Adapte le chemin selon ce que la cellule précédente montre
import glob

# Chercher le dossier train
train_candidates = glob.glob(f'{PLANTDOC_DIR}/**/train', recursive=True)
if not train_candidates:
    train_candidates = glob.glob(f'{PLANTDOC_DIR}/**/Train', recursive=True)

print(f"Dossiers train trouvés : {train_candidates}")

if train_candidates:
    TRAIN_DIR = train_candidates[0]
    plantdoc_classes = sorted(os.listdir(TRAIN_DIR))
    print(f"\n{len(plantdoc_classes)} classes PlantDoc :")
    for c in plantdoc_classes:
        n = len(os.listdir(os.path.join(TRAIN_DIR, c)))
        print(f"  {n:4d} images  {c}")
```

```python
import json

# Charger la liste des classes de ton modèle
with open(f'{PROJECT_ROOT}/results/class_names.json') as f:
    my_classes = json.load(f)

print(f"Tes {len(my_classes)} classes :")
for i, c in enumerate(my_classes):
    print(f"  {i:2d}  {c}")
```

```python
# Mapping PlantDoc → tes classes (basé sur l'analyse manuelle)
plantdoc_to_my_class = {
    "Apple_Scab_Leaf":                       "Apple___Apple_scab",
    "Apple_leaf":                            "Apple___healthy",
    "Apple_rust_leaf":                       "Apple___Cedar_apple_rust",
    "Bell_pepper_leaf":                      "Pepper,_bell___healthy",
    "Bell_pepper_leaf_spot":                 "Pepper,_bell___Bacterial_spot",
    "Blueberry_leaf":                        "Blueberry___healthy",
    "Cherry_leaf":                           "Cherry___healthy",
    "Corn_Gray_leaf_spot":                   "Corn___Cercospora_leaf_spot Gray_leaf_spot",
    "Corn_leaf_blight":                      "Corn___Northern_Leaf_Blight",
    "Corn_rust_leaf":                        "Corn___Common_rust",
    "Peach_leaf":                            "Peach___healthy",
    "Potato_leaf_early_blight":              "Potato___Early_blight",
    "Potato_leaf_late_blight":               "Potato___Late_blight",
    "Raspberry_leaf":                        "Raspberry___healthy",
    "Soyabean_leaf":                         "Soybean___healthy",
    "Squash_Powdery_mildew_leaf":            "Squash___Powdery_mildew",
    "Strawberry_leaf":                       "Strawberry___healthy",
    "Tomato_Early_blight_leaf":              "Tomato___Early_blight",
    "Tomato_Septoria_leaf_spot":             "Tomato___Septoria_leaf_spot",
    "Tomato_leaf":                           "Tomato___healthy",
    "Tomato_leaf_bacterial_spot":            "Tomato___Bacterial_spot",
    "Tomato_leaf_late_blight":               "Tomato___Late_blight",
    "Tomato_leaf_mosaic_virus":              "Tomato___Tomato_mosaic_virus",
    "Tomato_leaf_yellow_virus":              "Tomato___Tomato_Yellow_Leaf_Curl_Virus",
    "Tomato_mold_leaf":                      "Tomato___Leaf_Mold",
    "Tomato_two_spotted_spider_mites_leaf":  "Tomato___Spider_mites Two-spotted_spider_mite",
    "grape_leaf":                            "Grape___healthy",
    "grape_leaf_black_rot":                  "Grape___Black_rot",
}

# Vérifier que toutes les cibles existent bien dans mes 38 classes
unknown = [v for v in plantdoc_to_my_class.values() if v not in my_classes]
assert not unknown, f" Classes cibles inconnues : {unknown}"
print(f" Mapping valide : {len(plantdoc_to_my_class)} classes PlantDoc → {len(set(plantdoc_to_my_class.values()))} de tes 38 classes")

# Compter combien d'images terrain on apporte par classe (train uniquement)
from collections import defaultdict

train_count = defaultdict(int)
test_count = defaultdict(int)

for pd_class, my_class in plantdoc_to_my_class.items():
    train_dir = os.path.join(TRAIN_DIR, pd_class)
    test_dir = os.path.join(PLANTDOC_DIR, 'test', pd_class)
    if os.path.exists(train_dir):
        train_count[my_class] += len(os.listdir(train_dir))
    if os.path.exists(test_dir):
        test_count[my_class] += len(os.listdir(test_dir))

print(f"\n Images PlantDoc apportées par classe (train | test) :")
for c in sorted(set(train_count) | set(test_count)):
    idx = my_classes.index(c)
    print(f"  [{idx:2d}] {c:55s}  train={train_count[c]:4d}  test={test_count[c]:4d}")

total_train = sum(train_count.values())
total_test = sum(test_count.values())
print(f"\n  TOTAL train : {total_train}")
print(f"  TOTAL test  : {total_test}")
print(f"  Classes sans renfort terrain : {38 - len(set(plantdoc_to_my_class.values()))}/38")
```

```python
import json

mapping_path = f'{PROJECT_ROOT}/results/plantdoc_mapping.json'
with open(mapping_path, 'w', encoding='utf-8') as f:
    json.dump(plantdoc_to_my_class, f, ensure_ascii=False, indent=2)

print(f" Mapping sauvegardé : {mapping_path}")
```
